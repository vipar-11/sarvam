## Write a function to handle a tool call

The code below:

1. Interepts the message from the LLM requesting a tool call
2. Invokes the tool
3. Records the invocation of the tool and tool response into the chat history 

```python
def handle_tool_call(tool_call: any, messages: list):
    f_name = tool_call.function.name
    f_args = json.loads(tool_call.function.arguments)
    
    result = invoke_tool(f_name, f_args)
    
    messages.append(
        {
            "role": "assistant",
            "tool_calls": [
                {
                    "id": tool_call.id,
                    "type": "function",
                    "function": {
                        "name": tool_call.function.name,
                        "arguments": tool_call.function.arguments,
                    },
                }
            ],
        }
    )
    
    messages.append(
        {
            "role": "tool",
            "tool_call_id": tool_call.id,
            "content": str(result),
        }
    ) 
```
