## 2. Pipeline Coordinator

```python
def process_voice_pipeline(audio_path: str):
    if not audio_path:
        return (
            "No audio recorded.",
            "Awaiting audio input...",
            None
        )

    # 1. Audio -> Text
    transcribed_text = speech_to_text(audio_path)

    # 2. Text -> LLM Response
    bot_response_text = chat_with_bot(transcribed_text)

    # 3. LLM Response -> Audio (Echoes recorded audio by default)
    output_audio = text_to_speech(bot_response_text, fallback_audio=audio_path)

    return transcribed_text, bot_response_text, output_audio
```

## 3. Launch Gradio Interface

```python
with gr.Blocks(title="Voice Agent") as demo:
    gr.Markdown("# 🎙️ Voice Agent Interface")
    gr.Markdown("Record audio via microphone and click **Process Audio** to test the pipeline (Echo mode active by default).")

    with gr.Row():
        with gr.Column(scale=1):
            audio_input = gr.Audio(
                sources=["microphone"],
                type="filepath",
                label="1. Record Voice"
            )
            process_btn = gr.Button("▶️ Process Audio", variant="primary")

        with gr.Column(scale=1):
            stt_output = gr.Textbox(label="A. Transcribed Speech (STT)", interactive=False)
            bot_output = gr.Textbox(label="B. Chatbot Response (LLM)", interactive=False)
            audio_output = gr.Audio(label="C. Audio Playback (TTS Response / Echo)", type="filepath", interactive=False)

    process_btn.click(
        fn=process_voice_pipeline,
        inputs=[audio_input],
        outputs=[stt_output, bot_output, audio_output]
    )

demo.launch()
```
