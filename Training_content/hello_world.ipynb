```python
from dotenv import load_dotenv
import os
```

## Setup API Keys

```python
load_dotenv()
SARVAM_API_KEY = os.getenv('SARVAM_API_KEY')
```

## Make your first API Call

```python
from sarvamai import SarvamAI
```

```python
client = SarvamAI(
    api_subscription_key=SARVAM_API_KEY,
)

response = client.chat.completions(
    model="sarvam-105b-conversations",
    messages=[
        {"role": "user", "content": "My name is Jayaram"},
        {"role": "assistant", "content": "Nice to meet you, Jayaram. What would you like to talk about today?"},
        {"role": "user", "content": "ok, Do you rememeber my name ?"},
    ],
)
print(response.choices[0].message.content)
```

```python
system_prompt = """
You are a helpful Customer Services Rep working for ICICI Bank
"""
```

```python
model="sarvam-105b-conversations",
messages=[
    {"role": "system", "content": system_prompt},
]
```

```python
user_message = ''
while user_message != 'quit':
    user_message = input('\nUser:')
    messages.append(
        {"role": "user", "content": user_message.strip()}
    )
    response = client.chat.completions(
        model="sarvam-105b-conversations",
        messages=messages
    )
    print('\nAssistant: ', response.choices[0].message.content.strip())
    messages.append(
        {"role": "assistant", "content": response.choices[0].message.content}
    )
```

## Showcasing Temperature

```python
response = client.chat.completions(
    model="sarvam-105b-conversations",
    messages=[
        {"role": "user", "content": "tell me a joke"},
    ],
    temperature=0.5
)
print(response.choices[0].message.content)
```

## Streaming API

```python
response = client.chat.completions(
    model="sarvam-105b-conversations",
    messages=[
        {"role": "user", "content": "Tell me about Sarvam AI"},
    ],
    temperature=0.5
)
print(response.choices[0].message.content)
```

```python
stream = client.chat.completions(
    model="sarvam-105b-conversations",
    messages=[
        {"role": "user", "content": "Tell me about Sarvam AI"},
    ],
    temperature=0.5,
    stream=True,
)

for chunk in stream:
    # The final chunk reports usage and has no choices — guard before indexing
    if chunk.choices and chunk.choices[0].delta.content:
        print(chunk.choices[0].delta.content, end="", flush=True)
```
