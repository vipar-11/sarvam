# Speech to Text

```python
from sarvamai import SarvamAI
```

```python
from dotenv import load_dotenv
import os
```

## Setup API Keys

```python
load_dotenv()
SARVAM_API_KEY = os.getenv('SARVAM_API_KEY')
```

```python
client = SarvamAI(api_subscription_key=SARVAM_API_KEY)
```

## Transcribe wav file

```python
def transcribe_audio_to_text(audio_data:any):
    response = client.speech_to_text.transcribe(    
        file=audio_data,
        model="saaras:v3",
        mode="transcribe"  # or "translate", "verbatim", "translit", "codemix"
    )
    return response.transcript
```

```python
transcribe_audio_to_text(open("./data/audio/output1.wav", "rb"))
```

## Setup Interface to record Audio

```python
import gradio as gr
from scipy.io import wavfile
import io

def transcribe_audio_in_memory(audio_tuple):
    if audio_tuple is None:
        return "No audio recorded yet."    
    
    sample_rate, audio_data = audio_tuple
    
    # Convert raw data into in-memory wav 
    wav_io = io.BytesIO()
    wavfile.write(wav_io, sample_rate, audio_data)
    wav_io.seek(0)  # Rewind the pointer to the start of the virtual file
    wav_io.name = "audio.wav" 
    
    try:
        return transcribe_audio_to_text(wav_io)
        
    except Exception as e:
        return f"API Error: {str(e)}"
```

```python
# 5. Launch the inline Gradio Interface
demo = gr.Interface(
    fn=transcribe_audio_in_memory,
    inputs=gr.Audio(sources=["microphone"], type="numpy"),
    outputs=gr.Textbox(label="Sarvam AI Transcription Result"),
    title="In-Memory Sarvam AI Transcriber"
)

demo.launch(inline=True)
```

```python
demo.close()
```
