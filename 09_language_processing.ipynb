# Language Processing

```python
from sarvamai import SarvamAI
```

```python
from dotenv import load_dotenv
import os
```

## Setup API Keys

```python
load_dotenv()
SARVAM_API_KEY = os.getenv('SARVAM_API_KEY')
```

```python
client = SarvamAI(api_subscription_key=SARVAM_API_KEY)
```

## Detect Language

```python
#text = "வணக்கம், எப்படி இருக்கிறீர்கள்? நான் சர்வம் ஏஐ பயிற்சி வகுப்பில் கலந்துகொண்டிருக்கிறேன்."
text = "नमस्ते, आप कैसे हैं? मैं सर्वम AI की एक वर्कशॉप में शामिल हो रहा हूँ।"
```

```python
response = client.text.identify_language(
    input=text
)

detected_language, detected_script = response.language_code, response.script_code
detected_language, detected_script
```

## Translate to English

```python
response = client.text.translate(
    input=text,
    source_language_code=detected_language,
    target_language_code="en-IN",
)
translated_text = response.translated_text
translated_text
```

### Translate Back to Original Language

```python
response = client.text.translate(
    input=translated_text,
    source_language_code="en-IN",
    target_language_code=detected_language,
)
response.translated_text
```

## Modern vs Classic Colloquial Translation

```python
text = "Hello, How are you ? hope you are enjoying the workshop and excited to learn about Sarvam AI"
```

#### Modern Colloquial (Tamil)

```python
response = client.text.translate(
    input=text,
    source_language_code="en-IN",
    target_language_code="ta-IN",    
    model="mayura:v1",
    mode="modern-colloquial",
)
response.translated_text
```

#### Classic Colloquial (Tamil)

```python
response = client.text.translate(
    input=text,
    source_language_code="en-IN",
    target_language_code="ta-IN",
    model="mayura:v1",
    mode="classic-colloquial",
)
response.translated_text
```

#### Modern Colloquial (Hindi)

```python
text = "I am coming to your house tomorrow to study for the certification exam, would you be avaialble ?"
```

```python
response = client.text.translate(
    input=text,
    source_language_code="en-IN",
    target_language_code="hi-IN",
    model="mayura:v1",
    mode="modern-colloquial",
)
response.translated_text
```

#### Classic Colloquial (Hindi)

```python
response = client.text.translate(
    input=text,
    source_language_code="en-IN",
    target_language_code="hi-IN",
    model="mayura:v1",
    mode="classic-colloquial",
)
response.translated_text
```

#### Speaker Gender - Male / Female

```python
response = client.text.translate(
    input=text,
    source_language_code="en-IN",
    target_language_code="hi-IN",
    speaker_gender="Male",  # Male, Female
    model="mayura:v1",
    mode="classic-colloquial",
)
response.translated_text
```
