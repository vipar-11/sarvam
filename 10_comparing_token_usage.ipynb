# Compare Token Usage across LLMs for Indic Language Processing

```python
from dotenv import load_dotenv
import os
```

## Setup API Keys

```python
load_dotenv()

SARVAM_API_KEY = os.getenv('SARVAM_API_KEY')
assert SARVAM_API_KEY is not None, 'Could NOT load SARVAM_API_KEY from .env'    

GROQ_API_KEY = os.getenv("GROQ_API_KEY")
assert SARVAM_API_KEY is not None, 'Could NOT load GROQ_API_KEY from .env'   

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
assert SARVAM_API_KEY is not None, 'Could NOT load DEEPSEEK_API_KEY from .env'   
```

```python
from openai import OpenAI
```

## Helper Function to call LLM using Open AI Interface

```python
def call_llm(base_url: str, api_key: str, model: str, message: str):
    
    client = OpenAI(
        base_url=base_url,
        api_key=api_key,
    )

    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "user", "content": message},
        ],
    )
    
    print(response.choices[0].message.content, '\n')

    # print(response.usage.total_tokens)
    print(response.usage.model_dump_json(indent=2))
    
```

### Setup an Indic Language Prompt

```python
# message = "नमस्ते, आप कैसे हैं?"
message = "ஹலோ, நீங்க எப்படி இருக்கீங்க?"
```

### Answer using SarvamAI

```python
call_llm(
    "https://api.sarvam.ai/v1", 
    SARVAM_API_KEY, 
    "sarvam-105b-conversations",
    message,
)
```

### Answer using Groq

```python
call_llm(
    "https://api.groq.com/openai/v1", 
    GROQ_API_KEY, 
    "openai/gpt-oss-120b",
    message,
)
```

### Answer using Deep Seek

```python
call_llm(
    "https://api.deepseek.com", 
    DEEPSEEK_API_KEY, 
    "deepseek-v4-pro",
    message,
)
```
