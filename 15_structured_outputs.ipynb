# Convert Unstructured text into a Structured Object

```python
from dotenv import load_dotenv
import os
```

## Setup API Keys

```python
load_dotenv()

SARVAM_API_KEY = os.getenv('SARVAM_API_KEY')
assert SARVAM_API_KEY is not None, 'Could NOT load SARVAM_API_KEY from .env'    

GROQ_API_KEY = os.getenv("GROQ_API_KEY")
assert SARVAM_API_KEY is not None, 'Could NOT load GROQ_API_KEY from .env'   

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
assert SARVAM_API_KEY is not None, 'Could NOT load DEEPSEEK_API_KEY from .env'   
```

```python
from openai import OpenAI
```

## Open AI's responses.parse() Interface

```python
def call_llm(base_url: str, api_key: str, model: str, text: str, text_format: any):
    
    client = OpenAI(
        base_url=base_url,
        api_key=api_key,
    )

    response = client.responses.parse(
        model=model,
        input=[
            {"role": "system", "content": "Extract the information."},
            {"role": "user", "content": text},
        ],
        text_format=text_format,
    )
    
    return response.output_parsed    
```

### Define pydantic model

```python
from pydantic import BaseModel
```

```python
class IncidentTicket(BaseModel):
    ticket_id: str
    date_YYYYMMDD: str
    company_name: str
    customer_name: str
    user_id: str
    account_numbers: list[str]
    other_identifiers: list[str]
    application_function: str
```

### Create a fictitious ticket summary

```python
ticket_summary = """
INC7126761 - Financial Advisor - John Doe (id - jdoe123) from ABC Financial Services reporting slowness in the Options Trading Screen.
The screen did NOT load despite repeated attempts around 11:50 AM on 22nd Sep 2026.
Please investigate
""".strip()
```

### Extract structured output using Groq

```python
incident_ticket = call_llm(
    "https://api.groq.com/openai/v1", 
    GROQ_API_KEY, 
    "openai/gpt-oss-120b",
    ticket_summary,
    IncidentTicket
)
print(incident_ticket.model_dump_json(indent=2))
```

### Extract structured output using Deep Seek

```python
incident_ticket = call_llm(
    "https://api.deepseek.com", 
    DEEPSEEK_API_KEY, 
    "deepseek-v4-pro",
    ticket_summary,
    IncidentTicket
)
print(incident_ticket.model_dump_json(indent=2))
```
