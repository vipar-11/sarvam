# Showcasing Streaming Responses

```python
from dotenv import load_dotenv
import os
```

## Setup API Keys

```python
load_dotenv()
SARVAM_API_KEY = os.getenv('SARVAM_API_KEY')

assert SARVAM_API_KEY is not None, 'Could NOT load SARVAM_API_KEY from .env'    
```

## Long responses without streaming

```python
from sarvamai import SarvamAI

client = SarvamAI(
    api_subscription_key=SARVAM_API_KEY,
)
```

```python
response = client.chat.completions(
    model="sarvam-105b-conversations",
    messages=[
        {"role": "user", "content": "Tell me about Sarvam AI"},
    ],
)
print(response.choices[0].message.content)
```

## Long responses with streaming

```python
stream = client.chat.completions(
    model="sarvam-105b-conversations",
    messages=[
        {"role": "user", "content": "Tell me about Sarvam AI"},
    ],
    stream=True
)

for chunk in stream:
    if chunk.choices and chunk.choices[0].delta.content:
        print(chunk.choices[0].delta.content, end="", flush=True)
        if chunk.choices[0].finish_reason == "stop":
            break

#print(response.choices[0].message.content)
```
