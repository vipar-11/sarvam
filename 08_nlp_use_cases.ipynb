## Setup API Keys

```python
load_dotenv()
SARVAM_API_KEY = os.getenv('SARVAM_API_KEY')

assert SARVAM_API_KEY is not None, 'Could NOT load SARVAM_API_KEY from .env'    
```

## Use Case 1 - Sentiment Analysis

```python
from openai import OpenAI
```

```python
client = OpenAI(
    base_url="https://api.sarvam.ai/v1",
    api_key=SARVAM_API_KEY,
)
```

```python
feedback = [
    "The training was very useful and informative."
    "The concepts were explained clearly with practical examples,"
    "which helped me understand the topics better. I especially"
    "liked the hands-on activities and problem-solving sessions.",

    "Good to taking a full fo the section for us and we are all to "
    "thank you both sir and mam for your valuable time to taking the "
    "section for us and i great full to thank you",

    "Thanks for the valuable insights provided from your side "
    "and that being useful on our placement",

    "I was unable to follow the lectures, "
    "Using a portable voice amplifier would help us hear more clearly",

    "This program was good and useful for me",

    "Good explanation in every topic",
]
```

```python
system_prompt = """

Review the feedback from students of our workshop and classify as one of the following:

    VERY_POSITIVE
    POSITIVE
    NEUTRAL
    NEGATIVE
    VERY_NEGATIVE

Respond in exactly one word.
    
""".strip()
```

```python
def classify_feedback():
    for feedback_item in feedback:
        response = client.chat.completions.create(
            model="sarvam-105b-conversations",
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": feedback_item},
            ],
        )
    
        print(f'{feedback_item[:50]} ...', response.choices[0].message.content, '\n')
```

```python
classify_feedback()
```

## Use Case - Text Summarization

```python
system_prompt = """

Summarize the feedback received from students of our workshop in 2 or 3 lines.
    
""".strip()
```

```python
consolidated_feedback = "Consolidated Feedback:\n\n" + "\n\n".join(feedback)
print(consolidated_feedback)
```

```python
def summarize_feedback():
    response = client.chat.completions.create(
        model="sarvam-105b-conversations",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": consolidated_feedback},
        ],
    )

    print(response.choices[0].message.content)
    
```

```python
summarize_feedback()
```
