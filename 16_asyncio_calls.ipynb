# Sync vs AsyncIO - Performance Comparison

```python
from dotenv import load_dotenv
import os
```

## Setup API Keys

```python
load_dotenv()
SARVAM_API_KEY = os.getenv('SARVAM_API_KEY')

assert SARVAM_API_KEY is not None, 'Could NOT load SARVAM_API_KEY from .env'    
```

## Make a Sync Call to Sarvam AI

```python
from sarvamai import SarvamAI

client = SarvamAI(
    api_subscription_key=SARVAM_API_KEY,
)

response = client.chat.completions(
    model="sarvam-105b-conversations",
    messages=[
        {
            "role": "user", 
            "content": "Hi, How are you?"
        },        
    ],
)

print(response.choices[0].message.content)
```

## Make an Async Call to Sarvam AI

```python
from sarvamai import AsyncSarvamAI

client = AsyncSarvamAI(
    api_subscription_key=SARVAM_API_KEY,
)

response = await client.chat.completions(
    model="sarvam-105b-conversations",
    messages=[
        {
            "role": "user", 
            "content": "Hi, How are you?"
        },        
    ],
)

print(response.choices[0].message.content)
```

## Helper functions for Sync / Async Calls

```python
def ask_llm_sync(prompt: str):
    print(prompt)
    
    client = SarvamAI(
        api_subscription_key=SARVAM_API_KEY,
    )
    
    response = client.chat.completions(
        model="sarvam-105b-conversations",
        messages=[
            {
                "role": "user", 
                "content": prompt
            },        
        ],
    )
    
    return response.choices[0].message.content
```

```python
async def ask_llm_async(prompt: str):
    print(prompt)
    
    client = AsyncSarvamAI(
        api_subscription_key=SARVAM_API_KEY,
    )
    
    response = await client.chat.completions(
        model="sarvam-105b-conversations",
        messages=[
            {
                "role": "user", 
                "content": prompt
            },        
        ],
    )
    
    return response.choices[0].message.content
```

```python
ask_llm_sync("Hi ! How are you ?")
```

```python
await ask_llm_async("Hi ! How are you ?")
```

## Long Running LLM calls - Sync Version

```python
interview_questions = [
    "What is Linear Regression ?",
    "What are Precision and Recall ?",
    "What are some of the real-world applications of Benford's Distribution ?",
    "What is a Markov Chain ?",
    "Give me an example of technical challenge you have faced and how did you overcome it ?",
]
```

```python
import time
from IPython.display import Markdown, display

def ask_interview_questions_sync():
    
    start_time = time.perf_counter()
    answers = []
    for q in interview_questions:
        ans = ask_llm_sync(q)     
        answers.append(ans)
    end_time = time.perf_counter()
    time_taken = round(end_time - start_time)
    print(f'\n\nTook: {time_taken} seconds\n\n')
    
    display(Markdown( '\n\n'.join(answers) ))
    return time_taken
```

```python
sync_time = ask_interview_questions_sync()
```

## Long running LLM calls - Async Version

```python
import asyncio

async def ask_interview_questions_async():
    
    start_time = time.perf_counter()
    answers = await asyncio.gather(
        *[ask_llm_async(q) for q in interview_questions]
    )
    end_time = time.perf_counter()
    time_taken = round(end_time - start_time)
    print(f'\n\nTook: {time_taken} seconds\n\n')    
    display(Markdown( '\n\n'.join(answers) ))
    return time_taken
```

```python
async_time = await ask_interview_questions_async()
```

## % Time Savings

```python
savings = -100*(async_time/sync_time - 1)
print(f'Response Time reduced from {sync_time} to {async_time} seconds, Savings: {round(savings)}%')
```
