# Text to Speech

```python
from sarvamai import SarvamAI
from sarvamai.play import save, play
```

```python
from dotenv import load_dotenv
import os
```

## Setup API Keys

```python
load_dotenv()
SARVAM_API_KEY = os.getenv('SARVAM_API_KEY')
```

```python
client = SarvamAI(api_subscription_key=SARVAM_API_KEY)
```

## Convert Text to Voice

```python
#text = """How are you today ? Hope you are enjoying the course."""
text = "வணக்கம், எப்படி இருக்கிறீர்கள்? நான் Sarvam AI course நிறைவு செய்யப் போகிறேன்."
#text="அகர முதல எழுத்தெல்லாம் ஆதிபகவன் முதற்றே உலகு"
speaker = "shubh" # mani, shubh, priya
```

```python
audio = client.text_to_speech.convert(
    language_code="ta-IN",
    text=text,
    model="bulbul:v3",
    speaker=speaker,
    # pitch=0 #0.5, -0.5
    # pace=1.25
)
```

```python
play(audio, notebook=True)
```

## Save the audio

```python
 save(audio, "./data/audio/output1.wav")
```
