# Text to Speech

```python
from sarvamai import SarvamAI
from sarvamai.play import save, play
```

```python
from dotenv import load_dotenv
import os
```

## Setup API Keys

```python
load_dotenv()
SARVAM_API_KEY = os.getenv('SARVAM_API_KEY')
```

```python
client = SarvamAI(api_subscription_key=SARVAM_API_KEY)
```

```python
#text = """नमस्ते, क्या आपने अपने काम पूरे कर लिए हैं? जैसा कि कहा जाता है - 
#काल करे सो आज कर, आज करे सो अब। पल में परलय होएगी, बहुरि करेगा कब """

text = "इस परीक्षा के लिए मुझे पाँच दिन की छुट्टी मिली थी। मैं सर्कस, फिल्म या क्रिकेट देखने नहीं गया, बल्कि घर पर ही अच्छे से पढ़ाई की और अट्ठानबे अंक प्राप्त किए।"

language_code="hi-IN"
speaker = "simran_hi_conversation"
```

```python
audio = client.text_to_speech.convert(
    language_code=language_code,
    text=text,
    model="bulbul:v4-flash",
    speaker=speaker,
    pace=1.0
)
```

```python
play(audio, notebook=True)
```

## Save the audio

```python
 save(audio, "./data/audio/input_hindi.wav")
```
