# Text to Speech

```python
from sarvamai import SarvamAI
from sarvamai.play import save, play
```

```python
from dotenv import load_dotenv
import os
```

## Setup API Keys

```python
load_dotenv()
SARVAM_API_KEY = os.getenv('SARVAM_API_KEY')
```

```python
client = SarvamAI(api_subscription_key=SARVAM_API_KEY)
```

## Convert Text to Voice

```python
text = "Time management is very important for each of us. As they say - Time and tide wait for no man."
language_code="en-IN"
speaker = "sunny_en_social"
```

```python
# text="""வணக்கம், நேற்று கன்னியாகுமரி சென்று திருவள்ளுவர் சிலையை பார்த்தேன். திருவள்ளுவர் எழுதிய திருக்குறள்கள் மொத்தம் 1330."""

# language_code="ta-IN"
# speaker = "gokul_ta_narration"
```

```python
# text = """नमस्ते, क्या आपने अपने काम पूरे कर लिए हैं? जैसा कि कहा जाता है - 
# काल करे सो आज कर, आज करे सो अब। पल में परलय होएगी, बहुरि करेगा कब """

# language_code="hi-IN"
# speaker = "simran_hi_conversation"
```

```python
audio = client.text_to_speech.convert(
    language_code=language_code,
    text=text,
    model="bulbul:v4-flash",
    speaker=speaker,
    pace=1.0
)
```

```python
play(audio, notebook=True)
```

## Save the audio

```python
 save(audio, "./data/audio/output1.wav")
```
